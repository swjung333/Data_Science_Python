## Data Visualization

```python
# Histogram of Ratings

plt.figure(figsize=(8,5))
df_raw["Rating"].hist(bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Company Ratings")
plt.xlabel("Rating Value")
plt.ylabel("Number of Job Postings")  

plt.show()
```

```python
# Boxplot for Rating

plt.figure()
df_raw["Rating"].plot(kind="box", color="steelblue")
plt.title("Boxplot of Company Ratings")
plt.ylabel("Rating Value")

plt.show()
```

```python
# Bar chart for top 10 companies

top_title = df_raw["Job Title"].value_counts().head(10)

plt.figure(figsize=(8,5))
top_title.plot(kind="barh", color="steelblue", edgecolor="black") # 'h' stands for horizontal
plt.title("Top 10 most common Job Titles")
plt.xlabel("Number of Job Postings")
plt.tight_layout()

plt.show()
```

```python
top_companies = df_raw["Company Name"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_companies.plot(kind = "bar")
plt.xlabel("Comapny Name")
plt.ylabel("count")
plt.title("Top 10 Comapny names")
plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
top_companies = df_raw["Location"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_companies.plot(kind = "bar")
plt.xlabel("Location")
plt.ylabel("count")
plt.title("Top 10 Locations")
plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
# Pie chart for top 7 sectors

counts = df_raw["Sector"].value_counts().head(7)

plt.figure()
plt.pie(counts, labels=counts.index, autopct="%1.1f%%")
plt.title("Top 7 Sectors")

plt.show()
```
