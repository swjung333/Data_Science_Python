### Cleaning Data

```python
df = df_raw.copy()
df
```

```python
# Remove unnecessary columns
df = df.drop(["index", "Job Description", "Company Name", "Headquarters", "Founded", "Industry", "Competitors"], axis=1) # axis=1 removes columns, axis=0 removes rows
df.info()
```

```python
# Cleaning "Job Title"
df["Job Title"].value_counts()
```

```python
print(df["Job Title"].unique())
```

```python
# Remove Extra Space
df["Job Title"] = df["Job Title"].str.strip()
```

```python
import re #re means regular expression, it is for pattern matching and text manipulation

def get_level(title):
    title = title.lower()

    if (
        re.search(r"\bjr\.?\b", title) # \b is a word boundary, and words or anything after b is what we are looking for, and \. is for optional period, and ? is for optional character
         or "junior" in title
         or "early career" in title
         or "associate" in title
         or re.search(r"\bi\.?\b", title)
    ):
        return "Junior"
    elif (
        "sr" in title
        or "senior" in title
        or "lead" in title
        or "principal" in title
        or "Staff" in title
        or "expert" in title
        or "exprienced" in title
        or re.search(r"\bii\.?\b", title)
        or re.search(r"\biii\.?\b", title)

    ):
        return "senior"
    elif (
        "vp" in title
        or "vice president" in title
        or "director" in title
        or "chief" in title
    ):
        return "executive"
    else:
        return "mid"
    

df["Level"] = df["Job Title"].apply(get_level) # this is for applying the function to the column, and it will create a new column called "Level" with the results of the function
```

```python
def title_organizer(title):
    title = title.lower()




    # manager / lead (to categorize them)
    if (
        "manager" in title
        or "lead" in title
        or "vice president" in title
        or "director" in title
        or "chief" in title
        or "vp" in title
    ):
        return "manager"
    

    # machine learning
    elif(
        "machine learning" in title
        or "ml engineer" in title
        or "deep learning" in title
        or "ai/ml" in title
        or "ai" in title
        or "computer vision" in title
    ):
        return "ml engineer"
    
    # data engineer
    elif(
        "data engineer" in title
        or "big data" in title
        or "data architect" in title
        or "data integration" in title
        or "cloud data engineer" in title
    ):
        return "data engineer"
    
    # data analyst
    elif(
        "data analyst" in title
        or "business intelligence" in title
        or "analytics" in title
        or "analyst" in title
            
   ):
        return "data analyst"
    
    # data scientist
    elif(
        "data scientist" in title
        or "data science" in title
        or "decision scientist" in title
        or "ai data scientist" in title
            
   ):
        return "data scientist"
    
    # researcher
    elif(
        "research scientist" in title
        or "researcher" in title
        or "statistician" in title
        or "scientist" in title
            
   ):
        return "researcher"

    # software engineer
    elif(
        "software engineer" in title  
            
   ):
        return "software engineer"
    
    # other
    else:
        return "other" # even though we have one software engineer, we categorize it because it would be categorized as other, otherwise.
    
df["Job Title"] = df["Job Title"].apply(title_organizer)
```

```python
# To see the unique values in the "Salary Estimate" column
df["Salary Estimate"].value_counts()
```

```python
# remove (Glassdoor est.), (Emploer est.)

df["Salary Estimate"] = df["Salary Estimate"].str.replace("(Glassdoor est.)", "", regex=False)
df["Salary Estimate"] = df["Salary Estimate"].str.replace("(Employer est.)", "", regex=False)

df["Salary Estimate"] = df["Salary Estimate"].str.replace("$", " ", regex=False)
df["Salary Estimate"] = df["Salary Estimate"].str.replace("K", " ", regex=False)

df["Salary Estimate"] = df["Salary Estimate"].str.strip()


# split
df["Salary Estimate"] = df["Salary Estimate"].apply(lambda x: x.split()) # split the string into a list of words, and we will take the first and second element of the list to get the min and max salary
```

```python
# create new columns for min and max salary
df["Min_Salary"] = df["Salary Estimate"].apply(lambda x : x[0]) # 0 is the position of the first element of the list, because the first element is the minimum salary, and the second element is " - " which we don't need
df["Max_Salary"] = df["Salary Estimate"].apply(lambda x : x[2]) # 2 is the position of the third element of the list, because the second element is " - " which we don't need

df["Min_Salary"] = df["Min_Salary"].apply(lambda x:int(x)) # create a new column for the minimum salary by converting the string to integer
df["Max_Salary"] = df["Max_Salary"].astype(int) # another way to convert the string to integer for the maximum salary

df["Average_Salary"] = (df.Min_Salary + df.Max_Salary) / 2 # create a new column for the average salary by taking the average of the min and max salary
```
